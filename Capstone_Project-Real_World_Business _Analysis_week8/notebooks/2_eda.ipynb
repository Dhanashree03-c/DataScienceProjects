```python
#Import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("../data/cleaned_data.csv") #Load cleaned dataset
```

CHURN DISTRIBUTION

```python
#Churn distribution plot
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', data=df)   #Count of churn vs retained
plt.title("Customer Churn Distribution")
plt.show()
```

TENURE VS CHURN

```python
#Tenure distribution by churn
plt.figure(figsize=(8,5))
sns.boxplot(x='Churn', y='Tenure', data=df) #Compare tenure for churned vs retained
plt.title("Tenure Distribution by Churn")
plt.show()
```

MONTHLY CHARGES VS CHURN

```python
#Monthly charges distribution by churn
plt.figure(figsize=(8,5))
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)  #Compare charges for churned vs retained
plt.title("Monthly Charges vs Churn")
plt.show()
```

CONTRACT TYPE VS CHURN

```python
#Contract type vs churn
plt.figure(figsize=(8,5))
sns.countplot(x='Contract', hue='Churn', data=df)  #Count of churn by contract type
plt.xticks(rotation=45)
plt.title("Contract Type vs Churn")
plt.show()
```
